```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('compressed_data.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df['last review']=pd.to_datetime(df['last review'],errors='coerce')
```

```python
df.fillna({'review per month':0,'last review':df['last review'].min()},inplace=True)
```

```python
df.dropna(subset=['NAME','host name'],inplace=True)
```

```python
df.isna().sum()
```

```python
df=df.drop(columns=["license","house_rules"],errors='ignore')
```

```python
df.columns
```

```python
df.head()
```

```python
df['price']=df['price'].replace('[\$,]','',regex=True).astype(float)
df['service fee']=df['service fee'].replace('[\$,]','',regex=True).astype(float)
```

```python
df.head()
```

# Removing Duplicates

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

# Descriptive Statistics

```python
df.describe()
```

# Visualization

### Q1 : What is the distribution of listing prices?

```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=50,kde=True,color='red')
plt.title("Distribution of listing price")
plt.ylabel("Frequency")
```

# Conclusion
### The histogram shows a fairly even distribution of listing prices across different price ranges, indicating no particular concentration of listings in any specific price range. The KDE line helps visualize this even spread more clearly, confirming that the dataset contains listings with a wide variety of prices.

### Q2 : How are different room types distributed?

```python
df['room type']
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="room type" , data=df,color="hotpink")
plt.title("room type distribution")
plt.show()
              
```

### Q3 : How are listings distributed across different neighborhoods?

```python
plt.figure(figsize=(12,8))
sns.countplot(y='neighbourhood group',data=df,color="lightgreen",order=df['neighbourhood group'].value_counts().index)
plt.title('Number of listings by neighbourhood group')
plt.xlabel('Count')
plt.ylabel('neighbourhood Group')
plt.show()
```

### Q4 : What is the relationship between price and room type?

```python
df.head()
```

```python
plt.figure(figsize=(12,8))
sns.boxplot(x='room type',y='price',data=df)
plt.title("price vs room type")
plt.legend(title='Room Type')
plt.show()
```

### Q5 : How has the number of reviews changed over time?

```python

reviews_over_time=df.groupby(df['last review'].dt.to_period('M')).size()
plt.figure(figsize=(12,6))
reviews_over_time.plot(kind='line',color='red')
plt.title('Number of reviews over time')
plt.xlabel('Date')
plt.ylabel('Number of reviews')
plt.show()
```
